##### Multiplication Dot vs Element Wise(Hadamard)
Element Wise just do a direct broadcast and multiply it where as the dot product does 7 pattern
![image](../Jupyter%20Exercise/images/MatrixSlide.png)

```python
multiply_array1 = np.random.randint(0,10, size=(5,3))
multiply_array1
```

```python
multiply_array2 = np.random.randint(0,10, size=(5,3))
multiply_array2
```

```python
multiply_array1 * multiply_array2
```

#### Dot product 
![image](../Jupyter%20Exercise/images/DotMultiplicationSlide.png)

```python
np.dot(multiply_array1, multiply_array2)
```

```python
multiplicationAnswer = np.dot(multiply_array1, multiply_array2.T)
multiplicationAnswer
```

```python
weekly_sales_qty_raw = np.random.randint(5, 20, size=(5,3))
weekly_sales_qty_raw
```

```python
weekly_sales_qty = pd.DataFrame(weekly_sales_qty_raw, index=["Mon", "Tue", "Wed", "Thu", "Fri"], columns=["Almond Butter", "Hazelnut Butter", "Cashew Butter"])
weekly_sales_qty
```

```python
raw_price_chart = np.array([10, 8, 12])
raw_price_chart = raw_price_chart.reshape(1,3)
raw_price_chart.shape
```

```python
price_chart = pd.DataFrame(raw_price_chart, index=["Price"], columns=["Almond Butter", "Hazelnut Butter", "Cashew Butter"])
price_chart
```

```python
weekly_sales_qty["Total"] = np.dot(weekly_sales_qty, price_chart.T)
weekly_sales_qty
```
