#### Object Oriented API is preferred over pyplot API

```python
fig, ax = plt.subplots()
ax.plot(x, [50,100, 200, 250])
```

```python
plt.subplot(2, 1, 1)  # 2 rows, 1 column, first subplot
plt.plot([1, 2, 3])
plt.title('First Plot')

plt.subplot(2, 1, 2)  # 2 rows, 1 column, second subplot
plt.plot([4, 5, 6])
plt.title('Second Plot')

plt.tight_layout()
plt.show()
```

```python
fig, axs = plt.subplots(2, 1)  # 2 rows, 1 column
axs[0].plot([1, 2, 3])
axs[0].set_title('First Plot')
axs[1].plot([4, 5, 6])
axs[1].set_title('Second Plot')
fig.suptitle('My Two Subplots')
plt.tight_layout()
plt.show()
```

```python
heart_disease = pd.read_csv("heart-disease.csv")
heart_disease.head()
```

```python
heart_disease["age"].plot.hist()
```

```python
heart_disease.plot.hist(figsize=(10, 30), subplots=True);
```

```python
over_50 = heart_disease[heart_disease["age"] > 50]
over_50
```

```python
#Pyplot method
over_50.plot(kind= 'scatter', x = 'age', y = 'chol', c='target')
```

```python
## OO method(OO but not pure mixed with Pyplot)
fig, ax = plt.subplots(figsize=(10,6))
over_50.plot(kind='scatter', x='age', y = 'chol', c='target', ax=ax);
```

```python
## OO method from scratch
fig, ax = plt.subplots(figsize=(10, 6))
scatter = ax.scatter(x=over_50["age"], y=over_50["chol"], c=over_50["target"])
ax.set(title="Heart Disease and Chol levels", xlabel="Age", ylabel="Cholestrol")
# Add a legend
ax.legend(*scatter.legend_elements(), title="Target");
ax.axhline(over_50['chol'].mean(), linestyle='--');
```

```python
fig, ax = plt.subplots(figsize=(10,10), nrows=2, ncols=1)
scatter = ax[0].scatter(x=over_50["age"], y=over_50["chol"], c=over_50["target"]);
ax[0].set(title="Heart Disease and Cholestrol Level", ylabel="Cholestrol", xlabel="age");
legend1 = ax[0].legend(*scatter.legend_elements(), title="Target") #scatter = (handle, labels) in ax[0].legend(handles, labels, title="Target")
ax[0].axhline(y=over_50["chol"].mean(),linestyle="--")
ax[0].add_artist(legend1) #Mainly used if we need more than one legend

scatter2 = ax[1].scatter(x=over_50["age"], y=over_50['thalach'], c=over_50["target"])
ax[1].set(title="Heart Disease and High Heart Rate", ylabel="Max Heart Rate", xlabel="age")
legend2 = ax[1].legend(*scatter2.legend_elements(), title="Target")
ax[1].axhline(y=over_50["thalach"].mean(),linestyle="--")
```

```python
plt.style.use('seaborn-v0_8-whitegrid')
fig, ax = plt.subplots(figsize=(10,9), nrows=2 )
scatter1 = ax[0].scatter(x=over_50["age"], y=over_50["chol"], cmap="winter", c=over_50["target"])
ax[0].set(title="Heart Disease & Cholestrol Levels", xlabel="Age", ylabel="Chol");
ax[0].legend(*scatter1.legend_elements(), title="Target");
ax[0].axhline(over_50["chol"].mean(), linestyle="--")

scatter2 = ax[1].scatter(x=over_50["age"], y=over_50["thalach"], cmap="winter", c=over_50["target"])
ax[1].set(title="Age & Max Heart Rate", xlabel="Age", ylabel="Max Heart Rate");
ax[1].legend(*scatter2.legend_elements(), title="Target");
ax[1].axhline(over_50["thalach"].mean(), linestyle="--")
```
